```python
import pandas as pd
import numpy as np
from pathlib import Path
from IPython.display import display
```

```python

def explore_table(path, head=5, top_values=5):
    """
    Quick profiling of a CSV/CSV.GZ table for Bronze -> Silver analysis.

    Shows:
      - sample rows
      - shape
      - inferred pandas dtypes
      - memory usage
      - null counts / percentages
      - duplicated rows
      - cardinality per column
      - candidate key columns
      - numeric summary
      - categorical summary
      - frequent values
    """

    path = Path(path)
    table_name = path.name.replace(".csv.gz", "").replace(".csv", "")

    print("=" * 100)
    print(f"TABLE: {table_name}")
    print("=" * 100)

    # low_memory=False avoids chunk-based mixed type inference warnings
    df = pd.read_csv(path, low_memory=False)

    # ------------------------------------------------------------------
    # 1. BASIC SAMPLE
    # ------------------------------------------------------------------
    print("\n[1] FIRST ROWS")
    display(df.head(head))

    # ------------------------------------------------------------------
    # 2. DIMENSIONS
    # ------------------------------------------------------------------
    print("\n[2] DIMENSIONS")
    print(f"Rows:    {len(df):,}")
    print(f"Columns: {df.shape[1]:,}")
    print(f"Cells:   {df.size:,}")

    memory_mb = df.memory_usage(deep=True).sum() / 1024**2
    print(f"Memory:  {memory_mb:,.2f} MB")

    # ------------------------------------------------------------------
    # 3. COLUMN TYPES
    # ------------------------------------------------------------------
    print("\n[3] COLUMN TYPES")

    types = pd.DataFrame({
        "column": df.columns,
        "dtype": df.dtypes.astype(str).values
    })

    display(types)

    # ------------------------------------------------------------------
    # 4. NULLS + CARDINALITY
    # ------------------------------------------------------------------
    print("\n[4] NULLS AND CARDINALITY")

    profile = pd.DataFrame({
        "dtype": df.dtypes.astype(str),
        "non_null": df.notna().sum(),
        "null_count": df.isna().sum(),
        "null_pct": (df.isna().mean() * 100).round(2),
        "unique_count": df.nunique(dropna=True),
    })

    profile["unique_pct"] = (
        profile["unique_count"] / len(df) * 100
    ).round(2)

    profile = profile.sort_values(
        ["null_pct", "unique_count"],
        ascending=[False, False]
    )

    display(profile)

    # ------------------------------------------------------------------
    # 5. DUPLICATES
    # ------------------------------------------------------------------
    print("\n[5] DUPLICATES")

    duplicated_rows = df.duplicated().sum()

    print(f"Duplicated complete rows: {duplicated_rows:,}")
    print(
        f"Duplicated complete rows (%): "
        f"{duplicated_rows / len(df) * 100:.4f}%"
        if len(df)
        else "N/A"
    )

    if duplicated_rows:
        print("\nExample duplicated rows:")
        display(df[df.duplicated(keep=False)].head(10))

    # ------------------------------------------------------------------
    # 7. NUMERIC SUMMARY
    # ------------------------------------------------------------------
    numeric_cols = df.select_dtypes(include=np.number).columns

    print("\n[7] NUMERIC SUMMARY")

    if len(numeric_cols):
        display(
            df[numeric_cols]
            .describe()
            .T
            .round(2)
        )
    else:
        print("No numeric columns.")

    # ------------------------------------------------------------------
    # 8. TEXT / CATEGORICAL SUMMARY
    # ------------------------------------------------------------------
    categorical_cols = df.select_dtypes(
        include=["object", "string", "category"]
    ).columns

    print("\n[8] TEXT / CATEGORICAL SUMMARY")

    categorical_summary = []

    for col in categorical_cols:
        non_null = df[col].dropna()

        categorical_summary.append({
            "column": col,
            "non_null": len(non_null),
            "unique": non_null.nunique(),
            "avg_length": (
                non_null.astype(str).str.len().mean()
                if len(non_null)
                else np.nan
            ),
            "max_length": (
                non_null.astype(str).str.len().max()
                if len(non_null)
                else np.nan
            )
        })

    if categorical_summary:
        display(pd.DataFrame(categorical_summary))
    else:
        print("No text/categorical columns.")

    # ------------------------------------------------------------------
    # 9. MOST COMMON VALUES
    # ------------------------------------------------------------------
    print("\n[9] MOST COMMON VALUES")

    for col in categorical_cols:
        # Do not dump effectively unique fields such as URLs or hashes
        unique_ratio = df[col].nunique(dropna=True) / max(len(df), 1)

        if unique_ratio <= 0.20:
            print(f"\n--- {col} ---")
            display(
                df[col]
                .value_counts(dropna=False)
                .head(top_values)
                .rename("count")
                .to_frame()
            )

    # ------------------------------------------------------------------
    # 10. POTENTIAL DATE COLUMNS
    # ------------------------------------------------------------------
    print("\n[10] POSSIBLE DATE COLUMNS")

    date_candidates = [
        col for col in df.columns
        if "date" in col.lower()
        or col.lower().endswith("_at")
    ]

    date_rows = []

    for col in date_candidates:
        parsed = pd.to_datetime(df[col], errors="coerce")

        date_rows.append({
            "column": col,
            "original_dtype": str(df[col].dtype),
            "parseable": parsed.notna().sum(),
            "unparseable_non_null": (
                df[col].notna() & parsed.isna()
            ).sum(),
            "min_date": parsed.min(),
            "max_date": parsed.max()
        })

    if date_rows:
        display(pd.DataFrame(date_rows))
    else:
        print("No obvious date columns detected.")


    print("\n" + "=" * 100)

    return df
```
